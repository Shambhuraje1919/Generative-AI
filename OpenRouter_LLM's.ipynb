```python
matching_prompt = '''You are an expert AI Job Match Analyzer. Your task is to analyze a candidate's resume and a provided job description (JD), then return a structured JSON object suitable for the "Match to Job Report" UI.

**CRITICAL INSTRUCTIONS:**
1. **Analyze Both Documents:** Thoroughly analyze both the candidate’s resume and the given job description text.
2. **Calculate Match Score:** Compute a final 'match_score' (an integer 0–100) that reflects the overall alignment between the resume and JD.
3. **Generate Breakdown:** Create a 'breakdown' array containing 4 major categories:
  - "Skills Match"
  - "Experience Match"
  - "Education Match"
  - "Keyword Match"
  For each category, include:
  * A numeric 'score' (0–100)
  * A short 'description' summarizing your evaluation.
4. **Identify Keyword Overlaps:**
  Analyze both texts and extract important keywords (skills, tools, responsibilities, etc.).
  Return a structured object 'keyword_overlap' with:
  * 'overlap': list of common keywords found in both
  * 'missing_in_resume': keywords from JD not found in resume
  * 'extra_in_resume': keywords found in resume but not in JD
5. **Generate AI Suggestions:**
  Create 2–5 specific and actionable AI suggestions under 'ai_suggestions'.
  Each suggestion must include:
  * 'title' – concise summary (e.g., "Add Missing Keywords")
  * 'original' – the weak/missing element from the resume
  * 'upgraded' – a rewritten or improved version addressing the issue
6. **Summarize the Job Description:**
  Write a short, human-readable 'jd_summary' (2–3 sentences) describing the job's key requirements and focus areas.
7. **JSON Format:**
  You MUST return only a valid JSON object matching the `JSON_RESPONSE_SCHEMA` below.
  Do not include any other text, greetings, or explanations.

**USER DATA PROVIDED:**
- Resume Text:
{resume_text}

- Job Description Text:
{jd_text}

**JSON_RESPONSE_SCHEMA (Your output MUST match this):**
{{
 "match_score": "integer (0-100)",
 "breakdown": [
  {{
   "title": "string (e.g., 'Skills Match')",
   "score": "integer (0-100)",
   "description": "string (brief summary of category evaluation)"
  }}
 ],
 "keyword_overlap": {{
  "overlap": ["string", "string"],
  "missing_in_resume": ["string", "string"],
  "extra_in_resume": ["string", "string"]
 }},
 "ai_suggestions": [
  {{
   "title": "string (e.g., 'Add Missing Keywords')",
   "original": "string (e.g., 'No mention of Agile or Scrum.')",
   "upgraded": "string (e.g., 'Include Agile or Scrum experience if applicable.')"
  }}
 ],
 "jd_summary": "string (2–3 sentences summarizing the job description)"
}}'''
```

```python
def run_openrouter_model(model_name, resume_text, jd_text):

    llm = ChatOpenAI(
        model=model_name,
        api_key=os.environ["OPENROUTER_API_KEY"],
        base_url="https://openrouter.ai/api/v1",
        temperature=0,
        max_tokens=3000,
    )

    prompt = f"""
  {matching_prompt}

CANDIDATE RESUME:
{resume_text}

JOB DESCRIPTION:
{jd_text}
"""

    start = time.time()

    response = llm.invoke(prompt)

    latency = time.time() - start

    print("=" * 70)
    print(f"MODEL: {model_name}")
    print(f"LATENCY: {latency:.2f} seconds")
    print("=" * 70)
    print(response.content)

    return response.content, latency
```

```python
import requests

headers = {
    "Authorization": f"Bearer {os.environ['OPENROUTER_API_KEY']}"
}

response = requests.get(
    "https://openrouter.ai/api/v1/models",
    headers=headers
)

response.raise_for_status()

models = response.json()["data"]

print("Available models:", len(models))
```

```python
free_models = []

for m in models:
    pricing = m.get("pricing", {})

    if pricing.get("prompt") == "0" and pricing.get("completion") == "0":
        free_models.append(m)

print("Free models:", len(free_models))

for m in free_models:
    print(m["id"])
```

```python
OPENROUTER_MODEL = "google/gemma-4-26b-a4b-it:free"

gemma26_result, gemma26_latency = run_openrouter_model(
    OPENROUTER_MODEL,
    resume_text,
    jd_text
)
```

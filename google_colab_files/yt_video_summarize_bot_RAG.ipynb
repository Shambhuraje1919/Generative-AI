```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap = 200)
chunks =  splitter.create_documents(transcript)
```

```python
embedding = HuggingFaceEmbeddings(
    model = 'all-MiniLM-L6-v2'
)
```

```python
from langchain_core import documents
vector_store = FAISS.from_documents(embedding = embedding,
                     documents = chunks)
```

```python
vector_store
```

```python
from re import search
retriver = vector_store.as_retriever(search_type = 'similarity', search_kwargs={"k":4})
```

```python
retriver
```

```python
retriver.invoke('why studenst are job less')
```

```python
llm = ChatGroq(
    model = 'llama-3.1-8b-instant' ,
    api_key= Groq_API
)
```

```python
llm.invoke('hi')
```

```python
prompt = PromptTemplate(
    template = """
      You are a helpful assistant.
      Answer ONLY from the provided transcript context.
      If the context is insufficient, just say you don't know.

      {context}
      Question: {question}
    """,
    input_variables=['context','question']
)
```

```python
question = 'which is the best carrer in upcoming future regarding IT market in 2030'
retrived_ans = retriver.invoke(question)
```

```python
retrived_ans
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrived_ans)
context_text
```

```python
final_prompt = prompt.invoke({'context': context_text, "question": question})
```

```python
final_prompt
```

```python
ans = llm.invoke(final_prompt)
print(ans.content)
```

```python
from langchain_core.runnables import RunnableParallel , RunnablePassthrough , RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
  context_text = "\n\n".join(doc.page_content for doc in retrived_ans)
  return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': retriver | RunnableLambda(format_docs),
    'question': RunnablePassthrough()
})
```

```python
parallel_chain.invoke('tcs')
```

```python
parser = StrOutputParser()
```

```python
chain = parallel_chain | prompt | llm | parser
```

```python
chain.invoke(' AI jobs of future')
```

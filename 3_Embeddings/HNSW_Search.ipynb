```python
from sentence_transformers import SentenceTransformer
import faiss
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
query = [" the python are used for the using and working on the Ai and ML data and creating the neural networks so we can make it" ,
             "there are the other langugaes are used for the ML and DL like java but the programing and other things which can br differ"]
```

```python
embeddings = model.encode(
    query,
    convert_to_numpy = True
)
```

```python
dimenstion = embeddings.shape[1]
```

```python
index = faiss.IndexHNSWFlat(
    dimenstion ,
    32
)
```

```python
index
```

```python
index.hnsw.efConstruction = 40
```

```python
index.add(embeddings)
```
